### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
# sumar precio de los articulos vendidos
ingreso_total1 = tienda['Precio'].sum()
ingreso_total2 = tienda2['Precio'].sum()
ingreso_total3 = tienda3['Precio'].sum()
ingreso_total4 = tienda4['Precio'].sum()

print("Ingreso total de la tienda 1:", ingreso_total1)
print("Ingreso total de la tienda 2:", ingreso_total2)
print("Ingreso total de la tienda 3:", ingreso_total3)
print("Ingreso total de la tienda 4:", ingreso_total4)
```

```python
# que tienda obtuvo mayores ventas?
if ingreso_total1 > ingreso_total2 and ingreso_total1 > ingreso_total3 and ingreso_total1 > ingreso_total4:
    print("La tienda 1 tiene el mayor ingreso total.")
elif ingreso_total2 > ingreso_total1 and ingreso_total2 > ingreso_total3 and ingreso_total2 > ingreso_total4:
    print("La tienda 2 tiene el mayor ingreso total.")
elif ingreso_total3 > ingreso_total1 and ingreso_total3 > ingreso_total2 and ingreso_total3 > ingreso_total4:
    print("La tienda 3 tiene el mayor ingreso total.")
else:
    print("La tienda 4 tiene el mayor ingreso total.")

# que tienda obtuvo las peores ventas?
if ingreso_total1 < ingreso_total2 and ingreso_total1 < ingreso_total3 and ingreso_total1 < ingreso_total4:
    print("La tienda 1 tiene el menor ingreso total.")
elif ingreso_total2 < ingreso_total1 and ingreso_total2 < ingreso_total3 and ingreso_total2 < ingreso_total4:
    print("La tienda 2 tiene el menor ingreso total.")
elif ingreso_total3 < ingreso_total1 and ingreso_total3 < ingreso_total2 and ingreso_total3 < ingreso_total4:
    print("La tienda 3 tiene el menor ingreso total.")
else:
    print("La tienda 4 tiene el menor ingreso total.")
```

# 2. Ventas por categoría

```python
# Contar productos por categoría en cada tienda
categorias_tienda1 = tienda['Categoría del Producto'].value_counts()
categorias_tienda2 = tienda2['Categoría del Producto'].value_counts()
categorias_tienda3 = tienda3['Categoría del Producto'].value_counts()
categorias_tienda4 = tienda4['Categoría del Producto'].value_counts()

print('Tienda 1')
display(categorias_tienda1.head(3))
print('Tienda 2')
display(categorias_tienda2.head(3))
print('Tienda 3')
display(categorias_tienda3.head(3))
print('Tienda 4')
display(categorias_tienda4.head(3))
```

```python
#Calculo de satisfaccion del cliente por tienda

satisfaccion_tienda1 = tienda['Calificación'].mean().round(2)
satisfaccion_tienda2 = tienda2['Calificación'].mean().round(2)
satisfaccion_tienda3 = tienda3['Calificación'].mean().round(2)
satisfaccion_tienda4 = tienda4['Calificación'].mean().round(2)
```

# 3. Calificación promedio de la tienda


```python
print("Satisfacción promedio de la tienda 1:", satisfaccion_tienda1)
print("Satisfacción promedio de la tienda 2:", satisfaccion_tienda2)
print("Satisfacción promedio de la tienda 3:", satisfaccion_tienda3)
print("Satisfacción promedio de la tienda 3:", satisfaccion_tienda4)
```

# 4. Productos más y menos vendidos

```python
# Productos mas vendidos
productos_mas_vendidos_tienda1 = tienda['Producto'].value_counts().head(5)
productos_mas_vendidos_tienda2 = tienda2['Producto'].value_counts().head(5)
productos_mas_vendidos_tienda3 = tienda3['Producto'].value_counts().head(5)
productos_mas_vendidos_tienda4 = tienda4['Producto'].value_counts().head(5)

print("Tienda 1:")
display(productos_mas_vendidos_tienda1)
print("Tienda 2:")
display(productos_mas_vendidos_tienda2)
print("Tienda 3:")
display(productos_mas_vendidos_tienda3)
print("Tienda 4:")
display(productos_mas_vendidos_tienda4)
```

```python
# Productos menos vendidos
productos_menos_vendidos_tienda1 = tienda['Producto'].value_counts().sort_values(ascending=True).head(5)
productos_menos_vendidos_tienda2 = tienda2['Producto'].value_counts().sort_values(ascending=True).head(5)
productos_menos_vendidos_tienda3 = tienda3['Producto'].value_counts().sort_values(ascending=True).head(5)
productos_menos_vendidos_tienda4 = tienda4['Producto'].value_counts().sort_values(ascending=True).head(5)

print("Tienda 1:")
display(productos_menos_vendidos_tienda1)
print("Tienda 2:")
display(productos_menos_vendidos_tienda2)
print("Tienda 3:")
display(productos_menos_vendidos_tienda3)
print("Tienda 4:")
display(productos_menos_vendidos_tienda4)
```

# 5. Envío promedio por tienda

```python
# Costo de envio por tienda
envio_promedio_tienda1 = tienda['Costo de envío'].mean().round(2)
envio_promedio_tienda2 = tienda2['Costo de envío'].mean().round(2)
envio_promedio_tienda3 = tienda3['Costo de envío'].mean().round(2)
envio_promedio_tienda4 = tienda4['Costo de envío'].mean().round(2)

print("Costo de envío promedio de la tienda 1:", envio_promedio_tienda1)
print("Costo de envío promedio de la tienda 2:", envio_promedio_tienda2)
print("Costo de envío promedio de la tienda 3:", envio_promedio_tienda3)
print("Costo de envío promedio de la tienda 4:", envio_promedio_tienda4)
```

# 6. Visualizacion de los datos obtenidos

```python
# Importo biblioteca para graficos
import matplotlib.pyplot as plt
```

```python
# Datos para graficar
tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
ingresos = [ingreso_total1, ingreso_total2, ingreso_total3, ingreso_total4]
satisfaccion = [satisfaccion_tienda1, satisfaccion_tienda2, satisfaccion_tienda3, satisfaccion_tienda4]

print(tiendas)
print(ingresos)
print(satisfaccion)
```

```python
# Gráfico ingresos por tienda

# Escalar ingresos: de unidades → a millones (sino da error en grafico)
ingresos_millones = [valor / 1_000_000 for valor in ingresos]

# gráfico
plt.figure(figsize=(8,5))
bars = plt.bar(tiendas, ingresos_millones, color='skyblue')
plt.title('Ingresos por Tienda')
plt.xlabel('Tienda')
plt.ylabel('Ingresos (en millones)')
plt.ylim(1000, 1200)  # Hacemos zoom para notar las diferencias
plt.grid(axis='y')

# Etiquetas
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 10, f'{yval:.1f}', ha='center', va='bottom')

plt.show()
```

```python
# grafico ventas por categoria de tienda 4

categorias_tienda4 = tienda4['Categoría del Producto'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    categorias_tienda4,                      # los valores (ventas)
    labels=categorias_tienda4.index,         # los nombres de las categorías
    autopct='%1.1f%%',                       # muestra el porcentaje
    startangle=140,                          # ángulo inicial para rotar el gráfico
    colors=plt.cm.Paired.colors,             # la paleta de colores
    wedgeprops={'edgecolor': 'white'},       # divide las porciones con blanco
    pctdistance=0.85                         # distancia al centro para escribir los porcentajes
)

plt.title('Ventas por Categoría - Tienda 4')
plt.show()
```

```python
# Grafico satisfaccion promedio por tienda


plt.figure(figsize=(8,5))
bars = plt.bar(tiendas, satisfaccion, color='mediumorchid')
plt.title('Satisfacción Promedio por Tienda')
plt.xlabel('Tienda\n Nota: El eje Y inicia en 3.9 para resaltar diferencias en la satisfacción.')
plt.ylabel('Satisfacción (sobre 5)')
plt.ylim(3.9, 4.1)  # <<--- Zoom para el rango 3.9 - 4.1
plt.grid(axis='y')

# Agregar valores arriba de las barras
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.005, f'{yval:.2f}', ha='center', va='bottom')

plt.show()
```

```python
# Grafico productos mas vendidos para las 4 tiendas:

# Crear figura grande con 4 subplots
fig, axs = plt.subplots(2, 2, figsize=(18, 12))  # 2 filas, 2 columnas

# Datos (lista de tuplas: datos, título, subplot)
productos = [
    (productos_mas_vendidos_tienda1, 'Tienda 1', axs[0,0]),
    (productos_mas_vendidos_tienda2, 'Tienda 2', axs[0,1]),
    (productos_mas_vendidos_tienda3, 'Tienda 3', axs[1,0]),
    (productos_mas_vendidos_tienda4, 'Tienda 4', axs[1,1])
]

# Graficar cada tienda
for data, titulo, ax in productos:
    bars = data.plot(kind='barh', ax=ax, color='forestgreen')
    ax.set_title(f'Top 5 Productos Más Vendidos - {titulo}')
    ax.set_xlabel('Cantidad de Ventas')
    ax.set_ylabel('Producto')
    ax.invert_yaxis()  # Producto más vendido arriba
    ax.grid(axis='x')

    # Agregar etiquetas de valor en las barras
    for container in ax.containers:
        ax.bar_label(container, fmt='%d', label_type='edge', padding=3)

    # Ajustar el eje X para hacer zoom automático
    min_value = data.min() * 0.9  # un 10% debajo del mínimo
    max_value = data.max() * 1.1  # un 10% arriba del máximo
    ax.set_xlim(min_value, max_value)

# Ajustar espacios
plt.tight_layout()
plt.show()

```

```python
# Grafico productos menos vendidos para las 4 tiendas:

# Datos: productos menos vendidos
productos_menos_vendidos_tienda1 = tienda['Producto'].value_counts(ascending=True).head(5)
productos_menos_vendidos_tienda2 = tienda2['Producto'].value_counts(ascending=True).head(5)
productos_menos_vendidos_tienda3 = tienda3['Producto'].value_counts(ascending=True).head(5)
productos_menos_vendidos_tienda4 = tienda4['Producto'].value_counts(ascending=True).head(5)

# Crear figura
fig, axs = plt.subplots(2, 2, figsize=(18, 12))  # 2 filas, 2 columnas

# Datos para graficar
productos_menos = [
    (productos_menos_vendidos_tienda1, 'Tienda 1', axs[0,0]),
    (productos_menos_vendidos_tienda2, 'Tienda 2', axs[0,1]),
    (productos_menos_vendidos_tienda3, 'Tienda 3', axs[1,0]),
    (productos_menos_vendidos_tienda4, 'Tienda 4', axs[1,1])
]

# Graficar
for data, titulo, ax in productos_menos:
    bars = data.plot(kind='barh', ax=ax, color='salmon')
    ax.set_title(f'Top 5 Productos Menos Vendidos - {titulo}')
    ax.set_xlabel('Cantidad de Ventas')
    ax.set_ylabel('Producto')
    ax.invert_yaxis()  # El menos vendido arriba
    ax.grid(axis='x')

    # Agregar etiquetas de valor
    for container in ax.containers:
        ax.bar_label(container, fmt='%d', label_type='edge', padding=3)

    # Hacer zoom automático
    min_value = data.min() * 0.9
    max_value = data.max() * 1.1
    ax.set_xlim(min_value, max_value)

# Ajustar espacios
plt.tight_layout()
plt.show()

```
